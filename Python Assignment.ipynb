## 1. Explain the basic data types available in Python with examples?
Python is a dynamically typed language, so a variable's type is decided automatically from the value assigned to it.- **int** — whole numbers, e.g. `x = 10`-
**float** — decimal numbers, e.g. `y = 3.14`-
**complex** — numbers with a real and imaginary part, e.g. `z = 2 + 3j`-
**str** — sequence of characters, e.g. `s = "Hello World"`-
**bool** — True or False, e.g. `flag = True`-
**list** — ordered, mutable collection, e.g. `fruits = ["apple", "banana"]`- 
**tuple** — ordered, immutable collection, e.g. `point = (3, 4)`-
**dict** — key-value pairs, e.g. `student = {"name": "Amit", "marks": 85}`- 
**set** — unordered collection of unique items, e.g. `s = {1, 2, 3}`- 
**NoneType** — absence of a value, e.g. `x = None`

```python
x = 10                 # int
y = 3.14                # float
z = 2 + 3j              # complex
s = "Hello World"       # str
flag = True             # bool
fruits = ["apple", "banana", "cherry"]   # list
point = (3, 4)          # tuple
student = {"name": "Amit", "marks": 85}  # dict
nums = {1, 2, 3}        # set
x2 = None               # NoneType

for v in [x, y, z, s, flag, fruits, point, student, nums, x2]:
    print(v, "->", type(v))
```

## 3. Describe Arithmetic Operators, Assignment Operators, Comparison Operators, Logical Operators and Bitwise Operators in detail with examples.
**Arithmetic Operators:** `+  -  *  /  //  %  **`

```python
a, b = 17, 4
print(a + b)   # 21
print(a - b)   # 13
print(a * b)   # 68
print(a / b)   # 4.25
print(a // b)  # 4
print(a % b)   # 1
print(a ** b)  # 83521
```

**Assignment Operators:** `=  +=  -=  *=  /=  //=  %=  **=`

```python
x = 10
x += 5   # 15
x -= 3   # 12
x *= 2   # 24
x //= 5  # 4
print(x)
```

**Comparison Operators:** `==  !=  >  <  >=  <=`

```python
p, q = 8, 12
print(p == q)  # False
print(p != q)  # True
print(p < q)   # True
print(p >= q)  # False
```

**Logical Operators:** `and  or  not`

```python
m, n = 5, 0
print(m > 0 and n > 0)  # False
print(m > 0 or n > 0)   # True
print(not(m > 0))       # False
```

**Bitwise Operators:** `&  |  ^  ~  <<  >>`

```python
i, j = 6, 3       # binary: 110, 011
print(i & j)   # 2
print(i | j)   # 7
print(i ^ j)   # 5
print(~i)      # -7
print(i << 1)  # 12
print(i >> 1)  # 3
```

## 4. Write Python code to determine whether the given string is a Palindrome or not using slicing.
**Palindrome Check using Slicing`s[::-1]` reverses a string using slicing. If it equals the original, the string is a palindrome.**

```python
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

word = input("Enter a string: ")
if is_palindrome(word):
    print(f"'{word}' is a Palindrome")
else:
    print(f"'{word}' is not a Palindrome")
```

## 5. Write Python program to add two matrices and also find the transpose of the resultant matrix.

```python
A = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]

B = [[9, 8, 7],
     [6, 5, 4],
     [3, 2, 1]]

rows, cols = len(A), len(A[0])

# Matrix addition
result = [[A[i][j] + B[i][j] for j in range(cols)] for i in range(rows)]

print("Sum of Matrices:")
for row in result:
    print(row)

# Transpose of the resultant matrix
transpose = [[result[i][j] for i in range(rows)] for j in range(cols)]

print("\nTranspose of the Sum:")
for row in transpose:
    print(row)
```

## 6.Write Pythonic code to find Mean, Variance and Standard Deviation for a list of numbers.

```python
def mean(data):
    return sum(data) / len(data)

def variance(data):
    m = mean(data)
    return sum((x - m) ** 2 for x in data) / len(data)

def std_dev(data):
    return variance(data) ** 0.5

numbers = [12, 15, 18, 22, 25, 30]

print("Mean:", mean(numbers))
print("Variance:", variance(numbers))
print("Standard Deviation:", std_dev(numbers))
```

*(This can also be done directly using the `statistics` module: `statistics.mean()`, `statistics.variance()`, `statistics.stdev()`.)*

## 7. Discuss the Relation Between Tuples & Lists, and Tuples & Dictionaries in details.
**Tuples vs Lists**- Both are ordered and support indexing/slicing, holding mixed data types.- Lists are **mutable**; tuples are **immutable**.- Lists use `[ ]`; tuples use `( )`.- Tuples are faster, use less memory, and can be used as dictionary keys / set elements (lists cannot).- Convert with `list(t)` and `tuple(l)`.

```python
t = (1, 2, 3)
l = list(t)      # [1, 2, 3]
l.append(4)
t2 = tuple(l)     # (1, 2, 3, 4)
print(l, t2)
```

**Tuples vs Dictionaries**- A dictionary stores key-value pairs; a tuple stores an ordered collection of values with no keys.- Dictionary values are accessed by key; tuple values are accessed by index.- Because tuples are immutable/hashable, they're often used as dictionary keys.- A list of 2-item tuples can be converted into a dictionary using `dict()`.

```python
pairs = [("a", 1), ("b", 2), ("c", 3)]
d = dict(pairs)         # {'a': 1, 'b': 2, 'c': 3}

coordinates = {(0, 0): "origin", (1, 1): "diagonal point"}
print(d)
print(coordinates[(0, 0)])
```

## 8. Illustrate the following Set methods with an example.a) intersection()  b) union()  c) issubset()  d) difference()  e) update()  f) discard()

**a) intersection()** — common elements of both sets.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A.intersection(B))   # {3, 4}
```

**b) union()** — all unique elements from both sets.

```python
print(A.union(B))   # {1, 2, 3, 4, 5, 6}
```

**c) issubset()** — checks if all elements of one set exist in another.

```python
C = {1, 2}
print(C.issubset(A))   # True
```

**d) difference()** — elements in the first set but not the second.

```python
print(A.difference(B))   # {1, 2}
```

**e) update()** — adds elements of another iterable in place.

```python
A.update({7, 8})
print(A)   # {1, 2, 3, 4, 7, 8}
```

**f) discard()** — removes an element if present; no error if absent.

```python
A.discard(7)
A.discard(100)   # no error even though 100 is not in the set
print(A)   # {1, 2, 3, 4, 8}
```

## 10. What is a tuple in Python? What are the different operations on a tuple in Python?
Tuples in Python and Their OperationsA tuple is an ordered, **immutable** collection of items enclosed in `( )`. Tuples can hold items of different data types and allow duplicates.

```python
t = (10, 20, 30, 40, 20)

print(t[1])          # Indexing -> 20
print(t[1:3])         # Slicing -> (20, 30)
print(t + (50, 60))   # Concatenation
print(t * 2)          # Repetition
print(20 in t)        # Membership -> True
print(len(t))         # Length -> 5
print(t.count(20))    # Count occurrences -> 2
print(t.index(30))    # Index of first occurrence -> 2
print(min(t), max(t), sum(t))   # aggregate functions
for item in t:        # Iteration
    print(item, end=" ")
```

## 12. WAP python code for this dataset
students = (
(&quot;Amit&quot;, 85, &quot;Math&quot;),
(&quot;Riya&quot;, 92, &quot;Science&quot;),
(&quot;Karan&quot;, 76, &quot;Math&quot;),
(&quot;Anita&quot;, 89, &quot;History&quot;),
(&quot;Sonal&quot;, 92, &quot;Science&quot;)
)
1. Find the student with the highest marks.
2. Extract all students who studied &quot;Science&quot;.
3. Count how many students scored above 80.
4. Convert this tuple of records into a dictionary grouped by subject.
5. Find the average marks of all students.

```python
students = (
    ("Amit", 85, "Math"),
    ("Riya", 92, "Science"),
    ("Karan", 76, "Math"),
    ("Anita", 89, "History"),
    ("Sonal", 92, "Science")
)
```

**1. Student with the highest marks**

```python
topper = max(students, key=lambda s: s[1])
print("Topper:", topper)
```

**2. Students who studied Science**

```python
science_students = [s for s in students if s[2] == "Science"]
print(science_students)
```

**3. Count of students who scored above 80**

```python
count_above_80 = sum(1 for s in students if s[1] > 80)
print("Students scoring above 80:", count_above_80)
```

**4. Convert the tuple of records into a dictionary grouped by subject**

```python
grouped = {}
for name, marks, subject in students:
    grouped.setdefault(subject, []).append((name, marks))

print(grouped)
```

**5. Average marks of all students**

```python
average_marks = sum(s[1] for s in students) / len(students)
print("Average marks:", average_marks)
```
