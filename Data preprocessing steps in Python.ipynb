#Data Preprocessing in Python

###Step 1: import Libraries

```python
import pandas as pd
import numpy as np
```

###Step 2: Create Sample Dataset

```python
data = {
    "Age": [22, 25, np.nan, 30, 35, 40, np.nan, 28],
    "Salary": [25000, 30000, 35000, np.nan, 50000, 60000, 65000, 45000],
    "Experience": [1, 2, 3, 4, 5, 6, 7, 8]
}

df = pd.DataFrame(data)

print("Original Dataset")
print(df)
```

#Data Cleaning
###(a) Missing Data
####Fill missing values using Mean

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print("After Filling Missing Values")
print(df)
```

###(b) Noisy Data (Binning Methhod)

```python
df["Salary_Bins"] = pd.cut(df["Salary"], bins=3)

print(df)
```

###(c) Noisy Data (Regression)

```python
from sklearn.linear_model import LinearRegression

X = df[["Experience"]]
y = df["Salary"]

model = LinearRegression()
model.fit(X, y)

predicted = model.predict(X)

print("Predicted Salary")
print(predicted)
```

###(d) Noisy Data (Clustering)

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=42)

df["Cluster"] = kmeans.fit_predict(df[["Salary","Experience"]])

print(df)
```

#2. Data Transformation
###(a) Normalization

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["Age","Salary","Experience"]] = scaler.fit_transform(
    df[["Age","Salary","Experience"]]
)

print(df)
```

###(b)Attribute Selection

```python
selected = df[["Salary","Experience"]]

print(selected)
```

###(c) Discretization

```python
from sklearn.preprocessing import KBinsDiscretizer

disc = KBinsDiscretizer(
    n_bins=3,
    encode='ordinal',
    strategy='uniform'
)

df["Salary_Discrete"] = disc.fit_transform(df[["Salary"]])

print(df)
```

###(d) Concept Hierarchy Generation

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0,0.3,0.6,1],
    labels=["Young","Adult","Senior"]
)

print(df)
```

#3. Data Reduction
###(a) Data Cube Aggregation

```python
print(df.groupby("Cluster")["Salary"].mean())
```

###(b) attribute Subset Selection




```python
subset = df[["Age","Salary"]]

print(subset)
```

###(c) Numerosity Reduction

```python
sample = df.sample(4)

print(sample)
```

###(d) Dimensionality Reduction (PCA)

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

reduced = pca.fit_transform(
    df[["Age","Salary","Experience"]]
)

print(reduced)
```
